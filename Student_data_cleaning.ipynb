```python
dt=dt.dropna(subset=["final_grade"]).reset_index(drop=True)
```

```python
dt["name"]=dt["name"].fillna("unknown")
dt["gender"] = dt["gender"].fillna("unknown")
```

```python
for col in ["parental_support","online_classes_taken"]:
    dt[col]=dt[col].fillna(dt[col].mode()[0])
```

```python
num_cols=["attendance_rate","study_hours_per_week","PreviousGrade","extracurricular_activities"]
```

```python
for col in num_cols:
    dt[col]=dt[col].fillna(round(dt[col].median()))
```
